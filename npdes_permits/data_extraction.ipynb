```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data_old = pd.read_csv('output/2026-2-18/site_data.csv')
data_new = pd.read_csv('output/2026-4-26/site_data.csv')

```

```python
# Find rows in new that are not in old based on 'Agency' and 'Facility_Name'
merged = data_new.merge(
    data_old[['Agency', 'Facility_Name']], 
    on=['Agency', 'Facility_Name'], 
    how='left', 
    indicator=True
)
new_rows = merged[merged['_merge'] == 'left_only'].drop('_merge', axis=1)
new_rows = new_rows.reset_index(drop=True)
print(f"New dataframe created with {len(new_rows)} rows")
new_rows
```

```python
new_rows.to_csv('output/2026-4-26/new_site_data.csv', index=False)
print(f"Saved {len(new_rows)} rows to output/2026-4-26/new_site_data.csv")
```
